```python
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
filepath1 = "dataset/indonesia-province-jml-penduduk.json"
df = gpd.read_file(filepath1)  #reading the json file
#show the first 3 rows
df.head(3)
```

```python
df.columns
```

```python
df.plot()
```

```python
df.set_index("Propinsi", inplace=True) # so you can get a specific row by its province
df.head(2)
```

```python
df.index
```

```python
jabar = df.loc["JAWA BARAT"]["geometry"]
series = gpd.GeoSeries(jabar)
ax = series.plot()
plt.title("Peta Jawa Barat")
ax
```

```python
fig, ax = plt.subplots(1, 1, figsize=(20, 10))
df.plot(column='Jumlah Penduduk', ax=ax, legend=True, cmap='hot')
```
